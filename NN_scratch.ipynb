```python
# OPTIMIZING THE WEIGHTS OF A MULTILEVEL NN

torch.manual_seed(42)
X = torch.tensor([
    [2.0, 3.0],
    [1.0, 4.0],
    [3.0, 1.0]
])

y = torch.tensor([[5.0],[6.0],[4.0]])

class NeuralNetwork(nn.Module):

    def __init__(self):
        super().__init__()

        self.layer1 = nn.Linear(2, 3)    # 2 inputs   3 neurons (outputs)
        self.layer2 = nn.Linear(3, 2)    # 3 inputs   2 neurons (outputs)
        self.layer3 = nn.Linear(2, 1)    # 2 inputs   1 neuron
        self.relu = nn.ReLU()

    def forward(self, x):

        x = self.relu(self.layer1(x))
        x = self.relu(self.layer2(x))
        x = self.layer3(x)

        return x


model = NeuralNetwork()
loss_fun = nn.MSELoss()
print("type of model ",type(model))
print("type of loss_fun ",type(loss_fun))
#INITIALIZE AN OPTIMIZER

optimizer = optim.Adagrad(model.parameters(),lr=0.01)

#Now start the optimization
#FORWARD PROPAGATION --> COMPUTE LOSS --> BACKPROPAGATION --> UPDATE THE WEIGHTS AND BIAS

for epoch in range(1000):

    #FORWARD PROPAGATION                   
    prediction = model(X)

    #COMPUTE THE LOSS
    loss  = loss_fun(prediction,y)
    optimizer.zero_grad()

    #BACKPROPAGATION
    loss.backward()

    #UPDATE THE WEIGHTS AND BIAS
    optimizer.step()

 
print("\nOptimized parameters:")

'''for name, param in model.named_parameters():
    print(name)
    print(param.data)'''

print("FINAL WEIGHTS: ",model.layer3.weight.data)
print("FINAL BIAS: ",model.layer3.bias.data)
```

```python
## LETS DO NUMBER CLASSIFICATION

import torch 
import torch.nn as nn
from torchvision import datasets, transforms
import matplotlib.pyplot as plt

from torch.utils.data import DataLoader

transform = transforms.ToTensor()

train_set = datasets.MNIST(root='data',download=True, train=True, transform=transform)
test_set = datasets.MNIST(root='data', train=False, download=True, transform=transform)

print(f"The length of train set is {len(train_set)}")
print(f"The length of test set is {len(test_set)}")

image = train_set[400][0]
image = image.squeeze()
print("After squeeze:", image.shape)
print("Minimum pixel value:", image.min())
print("Maximum pixel value:", image.max())
print(image[10:15, 10:15])

#plt.imshow(image, cmap="gray")
#plt.show()

class NeuralNetwork(nn.Module):

    def __init__(self):
        super().__init__()

        self.flatten = nn.Flatten()
        self.layer1 = nn.Linear(784, 128)
        self.layer2 = nn.Linear(128, 64)
        self.layer3 = nn.Linear(64, 10)

        self.relu = nn.ReLU()

    def forward(self, x):

        x = self.flatten(x)

        x = self.layer1(x)
        x = self.relu(x)

        x = self.layer2(x)
        x = self.relu(x)

        x = self.layer3(x)

        return x

model = NeuralNetwork()

train_loader = DataLoader(
    train_set,
    batch_size=64,
    shuffle=True
)


loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.Adagrad(
    model.parameters(),
    lr=0.01
) 

epochs=10
for epoch in range(epochs):
    total_loss = 0
    for images, labels in train_loader:

        predictions = model(images)
        loss = loss_fn(predictions,labels)
        optimizer.zero_grad()

        loss.backward()
        optimizer.step()
        total_loss += loss.item()
    avg_loss = total_loss/ len(train_loader)
    print(f"Epoch {epoch+1}, Loss = {avg_loss:.6f}")

        
        
```

```python
## HOW GOOD IS THE NETWORK WE JUST TRAINED?
#
# A falling training loss only says "we are fitting the data we were shown".
# The honest question is how the model does on the 10,000 test images it has
# never seen. Run this cell right after the training cell above.

from torch.utils.data import DataLoader

test_loader = DataLoader(test_set, batch_size=1000, shuffle=False)

model.eval()          # evaluation mode (no-op here, but essential once you add Dropout/BatchNorm)

correct = 0
total   = 0

with torch.no_grad():                  # we are not learning, so skip building the gradient graph
    for images, labels in test_loader:

        logits = model(images)              # raw scores, shape [batch, 10]
        preds  = logits.argmax(dim=1)       # index of the largest score = the predicted digit

        correct += (preds == labels).sum().item()
        total   += labels.size(0)

print(f"Test accuracy: {correct}/{total} = {100 * correct / total:.2f}%")
```

```python
## MNIST WITH A CNN  --  TWO FLAVOURS OF GRADIENT DESCENT: MOMENTUM vs AdaGrad
#
# Note on naming: AdaGrad and "SGD + momentum" are NOT rival algorithms. Both are
# plain gradient descent with an add-on bolted onto the update step, and both are
# equally mini-batch ("stochastic") -- the batching comes from the DataLoader, not
# from the optimizer. Section 6 spells out exactly how the two differ.
#
# Why a CNN instead of the flat 784 -> 128 -> 64 -> 10 network above?
# Flattening the image throws away the fact that neighbouring pixels belong
# together. A convolution slides a small learnable filter across the image, so
# it can learn "an edge here", "a curve there" -- and reuse that same filter at
# every position. Fewer weights, and a far better fit to how images work.

import torch
import torch.nn as nn
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
from sklearn.metrics import classification_report


# ------------------------------------------------------------------
# 1. Hyperparameters -- everything you might want to tweak, in one place
# ------------------------------------------------------------------
EPOCHS     = 3      # how many full passes over the 60,000 training images
BATCH_SIZE = 64     # how many images go into one weight update
LR         = 0.01   # learning rate = how big each update step is
MOMENTUM   = 0.9    # only used by the momentum variant (see section 6)
SEED       = 42

DEVICE = torch.device("mps" if torch.backends.mps.is_available() else "cpu")
print("Training on:", DEVICE)


# ------------------------------------------------------------------
# 2. Data
# ------------------------------------------------------------------
transform = transforms.ToTensor()   # PIL image -> float tensor in [0,1], shape [1, 28, 28]

train_set = datasets.MNIST(root="data", train=True,  download=True, transform=transform)
test_set  = datasets.MNIST(root="data", train=False, download=True, transform=transform)

# Shuffle the training set so every epoch sees batches in a different order.
# The test set needs no shuffling, and can use big batches since we store no gradients.
train_loader = DataLoader(train_set, batch_size=BATCH_SIZE, shuffle=True)
test_loader  = DataLoader(test_set,  batch_size=1000,       shuffle=False)


# ------------------------------------------------------------------
# 3. The network
# ------------------------------------------------------------------
class CNN(nn.Module):
    """Conv -> Pool -> Conv -> Pool -> Flatten -> Dense -> Dense"""

    def __init__(self):
        super().__init__()

        self.features = nn.Sequential(
            nn.Conv2d(1, 16, kernel_size=3, padding=1),    # [1,28,28]  -> [16,28,28]   16 learned filters
            nn.ReLU(),
            nn.MaxPool2d(2),                               # [16,28,28] -> [16,14,14]   halve the resolution

            nn.Conv2d(16, 32, kernel_size=3, padding=1),   # [16,14,14] -> [32,14,14]
            nn.ReLU(),
            nn.MaxPool2d(2),                               # [32,14,14] -> [32,7,7]
        )

        self.classifier = nn.Sequential(
            nn.Flatten(),                                  # [32,7,7]   -> [1568]
            nn.Linear(32 * 7 * 7, 128),
            nn.ReLU(),
            nn.Dropout(0.25),                              # zero 25% of activations while training, to curb overfitting
            nn.Linear(128, 10),                            # 10 raw scores ("logits"), one per digit
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x


# ------------------------------------------------------------------
# 4. Training -- the same four steps as the simple network above
# ------------------------------------------------------------------
def train(model, optimizer, loss_fn, epochs=EPOCHS):

    for epoch in range(epochs):
        model.train()            # dropout ON
        total_loss = 0

        for images, labels in train_loader:
            images, labels = images.to(DEVICE), labels.to(DEVICE)

            predictions = model(images)               # 1. forward pass
            loss = loss_fn(predictions, labels)       # 2. how wrong were we?

            optimizer.zero_grad()                     #    clear the previous step's gradients
            loss.backward()                           # 3. backpropagate
            optimizer.step()                          # 4. nudge the weights

            total_loss += loss.item()

        print(f"  epoch {epoch + 1}/{epochs}   train loss = {total_loss / len(train_loader):.4f}")


# ------------------------------------------------------------------
# 5. Evaluation -- collect predictions, then a confusion matrix
# ------------------------------------------------------------------
@torch.no_grad()                 # no gradients needed: faster, and much less memory
def evaluate(model):
    model.eval()                 # dropout OFF -- always do this before measuring

    all_labels, all_preds = [], []

    for images, labels in test_loader:
        images = images.to(DEVICE)

        logits = model(images)                    # shape [batch, 10]
        preds  = logits.argmax(dim=1).cpu()       # biggest score wins -> predicted digit

        all_labels.append(labels)
        all_preds.append(preds)

    y_true = torch.cat(all_labels)
    y_pred = torch.cat(all_preds)

    confusion = torch.zeros(10, 10, dtype=torch.long)   # confusion[true, predicted] = count
    for t, p in zip(y_true, y_pred):
        confusion[t, p] += 1

    return y_true, y_pred, confusion


def report(y_true, y_pred, confusion):
    """Every metric below is read straight off the confusion matrix."""

    total   = confusion.sum().item()
    correct = confusion.diag().sum().item()       # the diagonal is where true == predicted

    print(f"\n  Accuracy: {correct}/{total} = {100 * correct / total:.2f}%\n")

    tp           = confusion.diag().float()       # correct hits per digit
    predicted_as = confusion.sum(dim=0).float()   # column sums: how often we *said* each digit
    truly_was    = confusion.sum(dim=1).float()   # row sums:    how often each digit *actually* appeared

    precision = tp / predicted_as.clamp(min=1)    # "when I say 7, how often is it really a 7?"
    recall    = tp / truly_was.clamp(min=1)       # "of all the real 7s, how many did I catch?"
    f1        = 2 * precision * recall / (precision + recall).clamp(min=1e-12)   # harmonic mean of the two

    print("  digit   precision    recall        f1    support")
    for d in range(10):
        print(f"  {d:5d} {precision[d]:11.3f} {recall[d]:9.3f} {f1[d]:9.3f} {int(truly_was[d]):10d}")
    print(f"  macro {precision.mean():11.3f} {recall.mean():9.3f} {f1.mean():9.3f} {total:10d}")

    print("\n  Confusion matrix (row = true digit, column = predicted):\n")
    print("         " + "".join(f"{p:6d}" for p in range(10)))
    for t in range(10):
        print(f"  true {t}" + "".join(f"{confusion[t, p]:6d}" for p in range(10)))

    # Same numbers, one line, via sklearn -- worth comparing against the hand-rolled version above
    print("\n  sklearn's classification_report, for comparison:\n")
    print(classification_report(y_true.numpy(), y_pred.numpy(), digits=3))


# ------------------------------------------------------------------
# 6. Train the same architecture twice, with two variants of gradient descent
# ------------------------------------------------------------------
#
# Every optimizer below starts from the SAME gradient g that loss.backward()
# just computed. They differ only in how they turn g into an actual step:
#
#   plain gradient descent    theta <- theta - lr * g
#
#   ... with momentum         v     <- mu * v + g                  (running average of g)
#                             theta <- theta - lr * v
#
#   ... with AdaGrad          G     <- G + g**2                    (accumulated, per weight)
#                             theta <- theta - lr * g / (sqrt(G) + eps)
#
# So momentum changes the DIRECTION of the step, smoothing it over time so the
# update carries velocity through flat regions and stops zig-zagging across
# narrow valleys. AdaGrad instead changes the SIZE of the step, separately for
# every single weight: a weight that has accumulated a lot of gradient gets a
# smaller and smaller step, while a rarely-updated weight keeps taking big ones.
# The two ideas are orthogonal -- Adam simply applies both at once, which is why
# Adam is the usual default in practice.

loss_fn = nn.CrossEntropyLoss()   # takes raw logits + integer labels, applies softmax internally

variants = {
    "momentum   (SGD, momentum=0.9)": lambda params: torch.optim.SGD(params, lr=LR, momentum=MOMENTUM),
    "AdaGrad    (per-weight step size)": lambda params: torch.optim.Adagrad(params, lr=LR),
}

for name, make_optimizer in variants.items():

    torch.manual_seed(SEED)       # identical random starting weights, so the comparison is fair
    model = CNN().to(DEVICE)
    optimizer = make_optimizer(model.parameters())

    print(f"\n{'=' * 62}\n{name}\n{'=' * 62}")

    train(model, optimizer, loss_fn)
    report(*evaluate(model))
```
